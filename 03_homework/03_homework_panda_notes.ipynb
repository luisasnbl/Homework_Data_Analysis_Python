#Notes on Panda

```python
#Import panda
import numpy as np
import pandas as pd
```

##Basic data structures in pandas
###Two types of classes:
1. Series: one-dimensional labeled array holding data of any type
   (e.g., integers, strings)
2. DataFrame: two-dimensional data structure that holds data like a twodimension array/table with rows and columns

##Object creation

```python
#Create a Series:
s = pd.Series([1, 3, 5, np.nan, 6, 8])
s
```

```python
#Create a DataFrame
#passing a NumPy array with a datetime index using date_range() and labeled columns:
dates = pd.date_range("20130101", periods=6)
dates

df = pd.DataFrame(np.random.randn(6,4), index=dates, columns=list("ABCD"))
df
```

```python
#Create a DataFrame
#passing a dictionary of objects where the keys are the column labels and the values are the column values
df2 = pd.DataFrame(
    {
        "A": 1.0,
        "B": pd.Timestamp("20130102"),
        "C": pd.Series(1, index=list(range(4)), dtype="float32"),
        "D": np.array([3] * 4, dtype="int32"),
        "E": pd.Categorical(["test", "train", "test", "train"]),
        "F": "foo",
    }
)
print(df2)

df2.dtypes
```

##Viewing data

```python
#use DataFrame.head() and DataFrame.tail() to view the top and bottom rows of the frame:
print(df.head())

df.tail(3)
```

```python
#display DataFrame.index or DataFrame.columns
print(df.index)
print(df.columns)
```

```python
#return NumPy representation of underlying data with DataFrame.to_numpy() without index or column labels
df.to_numpy()
```

```python
#quick statistic summary of your data with describe()
df.describe()
```

```python
#transpose data
df.T
```

```python
#sort by an axis: DataFrame.sort_index()
df.sort_index(axis=1, ascending=False)
```

```python
#sort by values: DataFrame.sort_values()
df.sort_values(by="B")
```
